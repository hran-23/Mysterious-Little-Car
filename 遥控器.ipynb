```python
import serial
arduino = serial.Serial("COM3")
```

```python
from pynput.keyboard import Key, Listener

def on_press(key):
    try:
        #print(f"Alphanumeric key pressed: {key.char}")
        k=key.char
        a='0y0'
        if k=="w":
            a='100y100'
        elif k=="s":
            a='-100y-100'
        elif k=="a":
            a='-100y100'
        elif k=="d":
            a='100y-100'
        arduino.write(a.encode())
    except AttributeError:
        print(f"Special key pressed: {key}")

def on_release(key):
    #print(f"Key released: {key}")
    if key == Key.esc:
        # 停止监听
        return False

#on_press=on_press,
# 设置监听器监听按键事件
with Listener(on_release=on_release,on_press=on_press) as listener:
    print("2345678")
    listener.join()
    

```
